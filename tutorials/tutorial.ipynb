# Sample Similarity Explorer: Practical Application Tutorial

## 1. Goal and Biological Question
The goal of this tutorial is to demonstrate how to use the `sample-similarity-explorer` package to analyze biological sample relationships in a real-world dataset.

**Biological Question:** Can we distinguish different biological samples based on their gene expression similarity? Specifically, we want to verify if samples from the GEO dataset GDS6063 exhibit high global correlation, which is expected for samples within a curated dataset of similar biological conditions.

````python
import sys
sys.path.append("..")

import pandas as pd
from src.sample_similarity_explorer.similarity import compute_sample_similarity
````

## 3. Data Description
We are using the **GDS6063** dataset from the NCBI Gene Expression Omnibus (GEO).
- **Type:** Gene expression matrix (RNA-seq/Microarray).
- **Structure:** Rows represent genes, and columns represent 10 biological samples.
- **Size:** The parsed matrix contains **12,624 genes** (features).
- **Stats:** The values represent normalized continuous expression counts.

````python
# Load the real dataset
data_path = "../data/gds6063_expression.csv"
df = pd.read_csv(data_path, index_col=0) # Assuming the first column is the Gene ID

# Display basic stats as required by the assignment
print(f"Dataset Shape: {df.shape}")
print("\nFirst 5 rows of the expression matrix:")
display(df.head())

print("\nBasic Summary Statistics:")
display(df.describe())
````

## 4. Running the Similarity Analysis
We will compute the **Spearman correlation** between samples. Spearman is a good choice for biological data as it is less sensitive to outliers than Pearson.

````python
# Compute similarity
spearman_sim = compute_sample_similarity(df, metric="spearman")

# Display the result summary
print("Spearman Similarity Matrix (Summary of Results):")
display(spearman_sim)

# Save the results
output_path = "../data/similarity_results_summary.csv"
spearman_sim.to_csv(output_path)
print(f"\nFull matrix saved to {output_path}")
````

````python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
sns.heatmap(spearman_sim, annot=True, cmap="YlGnBu", fmt=".2f", vmin=0.5, vmax=1.0)
plt.title("Sample Similarity Heatmap (GDS6063)")
plt.xlabel("Samples")
plt.ylabel("Samples")
plt.show()
````

## 5. Conclusion and Biological Answer

### Summary of Results:
The computed 10x10 similarity matrix shows that all samples in the GDS6063 dataset exhibit a Spearman correlation coefficient between **0.78 and 0.82**. This indicates a strong positive monotonic relationship between the expression profiles of these samples.

### Biological Answer:
Based on the analysis, we can answer our biological question: Yes, the samples show a high degree of similarity, which confirms they belong to a consistent experimental group. 

### Final Conclusion:
The `Sample Similarity Explorer` successfully quantified the relationships between high-dimensional biological samples. This tool is valuable for **Quality Control (QC)**—helping researchers identify outliers that don't cluster with expected replicates—and for **Exploratory Data Analysis (EDA)** to understand global sample patterns before proceeding to differential expression analysis.
