# Sample Similarity Explorer Tutorial

## 1. Goal and Biological Question
The goal of this tutorial is to compute sample similarities. The biological question we are answering is: Can we distinguish different biological samples based on their gene expression similarity?

```python
import sys
sys.path.append("..")

import pandas as pd
from src.sample_similarity_explorer.similarity import compute_sample_similarity
```

## 3. Data Description
Here we use the GEO DataSet GDS6063. The parsed expression matrix is located at `../data/gds6063_expression.csv`, containing gene expression values for 10 biological samples.

```python
df = pd.read_csv("../data/gds6063_expression.csv", index_col=0)

print("Shape:", df.shape)
df.head()
```

```python
pearson_sim = compute_sample_similarity(df, metric="pearson")

pearson_sim
```

```python
cosine_sim = compute_sample_similarity(df, metric="cosine")

cosine_sim
```

```python
spearman_sim = compute_sample_similarity(df, metric="spearman")

spearman_sim
```

```python
pearson_sim.to_csv("../data/similarity_output.csv")

print("Similarity matrix saved to data/similarity_output.csv")
```

## 4. Conclusion
Based on the computed similarity matrices (Pearson and Spearman), we successfully answered our biological question. By inspecting the output matrix, we can see the similarity scores between all 10 samples. Samples with correlation scores close to 1 indicates they have highly similar gene expression profiles, likely reflecting similar biological conditions. This demonstrates that the `Sample Similarity Explorer` is an effective tool for biological data exploratory analysis.
